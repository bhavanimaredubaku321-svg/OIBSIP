Libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
```

Load Dataset

```python
df = pd.read_csv("E-Commerce-Customer.csv")
```

Inspect Structure

```python
print ("Shape:", df.shape)
print("\nColumns:")
print(df.columns)
print("\nFirst 5 rows:") 
display(df.head())
print("\nDataset Information:")
df.info()
```

Missing Values

```python
print("Missing values:")
print(df.isnull().sum())
print("\nDuplicate rows:",df.duplicated().sum())
```

Handle Missing/Inconsistent Data

```python
df = df.drop_duplicates()
numeric_cols = df.select_dtypes(include=np.number).columns
df[numeric_cols] = df[numeric_cols].fillna(df[numeric_cols].median())
print("Missing values after cleaning:")
print(df.isnull().sum())
```

Descriptive Statistics

```python
df.describe()
```

Average Purchase Value

```python
print("Average Purchase Value:",df["Total"].mean())
```

Customer Lifetime

```python
customer_lifetime = df.groupby("CustomerID")["Total"].sum()
print("Average Customer Lifetime Value:",customer_lifetime.mean())
```

Purchase Frequency

```python
purchase_frequency = df.groupby("CustomerID")["InvoiceNumber"].nunique()
print("Average Purchase Frequency:",purchase_frequency.mean())
```
